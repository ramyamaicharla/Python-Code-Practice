### Inheritance
**What is Inheritance?**

- Inheritance allows one class to acquire the properties and methods of another class.

- child class inherits attributes and methods from a parent class.

- Parent Class → Child Class

**Why Inheritance?**
    
   - Code Reusability
   - Easy Maintenance
   - Reduces Duplicate Code

```python
# syntax
class Parent:

    # Parent properties
    pass

class Child(Parent):

    # Child properties
    pass
```

```python
# example
class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def bark(self):
        print("boww..")

d=Dog()
d.sound()
d.bark()

# 3. Real-Time Example
#---------------------
# Parent Class: Animal
# Child Class: Dog

# The Dog inherits common properties of Animal.
```

#### Constructor in Inheritance

```python
class Animal:
    def __init__(self):
        print("Animal Constructor")

class Dog(Animal):
    pass

d=Dog()

# The parent constructor is called automatically.
```

#### Using super()
- super() is used to call the parent class constructor or methods

```python
class Animal:

    def __init__(self):
        print("Animal Constructor")

class Dog(Animal):

    def __init__(self):
        super().__init__()
        print("Dog Constructor")

d = Dog()
```

#### Method Overriding
- The child class can redefine a method of the parent class

```python
class Animal:

    def sound(self):
        print("Animal Sound")

class Dog(Animal):

    def sound(self):
        print("Dog Barks")

d = Dog()

d.sound()
```

**Types of Inheritance**
- 1.Single Inheritance
- 2.Multilevel Inheritance
- 3.Multiple Inheritance
- 4.Hierarchical Inheritance
- 5.Hybrid Inheritance

#### 1.Single Inheritance
- one parent---one child

```python
class A:
    pass

class B(A):
    pass
```

```python
# example
class A:

    def show(self):
        print("Class A")

class B(A):

    def display(self):
        print("Class B")

obj = B()
obj.show()
obj.display()
```

#### 2. Multiple Inheritance
- One Child → Multiple Parents

```python
# syntax
class A:
    pass

class B:
    pass

class C(A,B):
    pass
```

```python
# example
class A:
    def show(self):
        print("class A")

class B:
    def display(self):
        print("class B")

class C(A,B):
        pass

obj=C()
obj.show()
obj.display()
        
```

#### 2.Multilevel Inheritance
- Grandparent → Parent → Child

```python
# syntax
class A:
    pass

class B(A):
    pass

class C(B):
    pass
```

```python
class A:
    def show(self):
        print("class A")

class B(A):
    def display(self):
        print("class B")
        
class C(B):
    def print_data(self):
        print("class C")

obj=C()
obj.show()
obj.display()
obj.print_data()    
```

#### 4. Hierarchical Inheritance
- One Parent → Multiple Children

```python
class Animal:
    def eat(self):
        print("eating")

class Dog(Animal):
    def bark(self):
        print("bark")

class Cat(Animal):
    def meow(self):
        print("meow")

d=Dog()
d.eat()
d.bark()

c=Cat()
c.eat()
c.meow()
```

#### 5. Hybrid Inheritance

- Hybrid inheritance is a combination of two or more inheritance types.

```python
# example
      A
     / \
    B   C
     \ /
      D
```

#### Advantages
- Reuses existing code
- Reduces code duplication
- Easy to maintain
- Improves readability
- Supports code extension

#### Disadvantages
- Deep inheritance can make code harder to understand.
- Changes in the parent class may affect child classes.
- Improper design can increase complexity.

#### Difference Between Parent and Child Class
| Parent Class             | Child Class          |
| ------------------------ | -------------------- |
| Base class               | Derived class        |
| Provides properties      | Inherits properties  |
| Can exist independently  | Depends on parent    |
| Contains common features | Can add new features |


#### Common Programs
- ###### Employee Inheritance

```python
class Employee:
    def work(self):
        print("employee is working")

class Manager(Employee):
    def manager(self):
        print("manager is managing")

m=Manager()
m.work()
m.manager()
```

##### Student Inheritance

```python
class Student:
    def study(self):
        print("student is studying")

class CollegeStudent(Student):
    def exam(self):
        print("writing exam")

s=CollegeStudent()
s.study()
s.exam()
```

##### Vehicle Inheritance

```python
class Vehicle:

    def start(self):
        print("Vehicle Started")

class Car(Vehicle):

    def drive(self):
        print("Car is Driving")

c = Car()
c.start()
c.drive()
```
